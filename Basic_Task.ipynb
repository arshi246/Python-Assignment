## Task 1: Cinema Ticket Price

### Task Description

User ki age input lein aur age ke according cinema ticket price display karein.

### Rules

- Age below 5 or above 100: Free
- Age 5–12: Rs. 300
- Age 13–59: Rs. 700
- Age 60-100: Rs. 400

### Requirements

1. User se name aur age input lein.
2. Age ke according ticket price calculate karein.
3. Customer name aur final ticket price display karein.
### Example Output

```
Enter your name: Ali
Enter your age: 10

Customer: Ali
Ticket Price: Rs. 300
```

````python
user_name = input("Enter your name : ").strip().capitalize()
age = int(input("Enter your age : "))

if age<5  or age> 100:
          ticket_price = "Free"
          print(f"Customer : {user_name}")
          print(f"Ticket Price : Rs.{ticket_price}")

elif age>=5 and age<=12:
          ticket_price = 300
          print(f"Customer : {user_name}")
          print(f"Ticket Price : Rs.{ticket_price}")

elif age >=13 and age<=59:
          ticket_price = 700
          print(f"Customer : {user_name}")
          print(f"Ticket Price : Rs.{ticket_price}")

elif age >=60 and age<=100:
          ticket_price = 400
          print(f"Customer : {user_name}")
          print(f"Ticket Price : Rs.{ticket_price}")

else:
          print("Invalid Input")
````

## Task 2: Temperature Converter

### Task Description

User se Celsius temperature input lein aur usko Fahrenheit mein convert karein. Temperature ka weather message bhi display karein.

### Formula

```
Fahrenheit = (Celsius × 9/5) + 32
```

### Weather Rules

- 10°C–19°C: Cold
    - Below 10°C: Very Cold
- 20°C–29°C: Normal
- 30°C or above: Hot

### Requirements
1. Celsius temperature input lein.
2. Fahrenheit calculate karein.
3. Weather condition display karein.

````python
try:
  celsius_temperature = float(input("Enter Temperature : "))
  Fahrenheit = (celsius_temperature * 9/5) + 32

  print(f"Celsius : {celsius_temperature}°C")
  print(f"Farenheit : {Fahrenheit}°C")

  if celsius_temperature < 10 and celsius_temperature <= 19:
    print(f"Weather : Very Cold")

  elif celsius_temperature >= 10 and celsius_temperature <= 19:
    print("Weather : Normal")

  elif celsius_temperature >= 20 and celsius_temperature <= 29:
    print("Weather : Cold")

  elif celsius_temperature >= 30:
    print("Weather : Hot")

except ValueError:
    print("Invalid Input.Please Enter Numeric Temperature")
````

## Task 3: Mobile Recharge Calculator

### Task Description

User se recharge amount input lein. Company 3% tax deduct karti hai. Calculate karein ke customer ko kitna final balance milega.

### Formula

```
Tax = Recharge Amount × 3 / 100
Final Balance = Recharge Amount - Tax
```

### Requirements

1. Mobile number input lein.
2. Recharge amount input lein.
3. Tax calculate karein.
4.Final received balance display karein.

````python
try:
  mob_num = input("Enter your Mobile Number : ")
  if len(mob_num) == 11:
    recharge_amount = float(input("Enter Recharge Amount : "))
    tax = recharge_amount * 3 / 100
    final_balance = recharge_amount - tax

    print(f"Mobile Number : {mob_num}")
    print(f"Recharge Amount : {recharge_amount}")
    print(f"Tax : {tax}")
    print(f"Final Balance : {final_balance}")
  else:
    print("Enter Correct Number")

except ValueError:
  print("Invalid Input .Please Enter Numeric Number")
````


## Task 4: Weekly Steps Tracker

### Task Description

User se seven days ke walking steps input lein. Total aur average steps calculate karein.

### Requirements

1. `for` loop seven times run karein.
2. Har din ke steps input lein.
3. Total steps calculate karein.
4. Daily average calculate karein.
5. Agar average `5,000` ya above ho to `Active Week` display karein, warna `Needs More Activity`.

### Example Output

```
Total Steps: 42000
Daily Average: 6000
Status: Active Week



````python
total_steps = 0
for i in range(1,8):
  walking_steps = int(input("Enter Walking Steps : "))
  total_steps += walking_steps

average = total_steps/7

print(f"Total Steps : {total_steps}")
print(f"Average : {average}")

if average >= 5000:
  print("Active Week")
else:
  print("Needs More Activity50")
````

Task 5: Secure PIN Checker


Task Description


Ek secret PIN program mein save karein. User ko correct PIN enter karne ke liye maximum three attempts dein.


Requirements :

Secret PIN 1234 save karein.


User se PIN input lein.


Correct PIN par Access Granted display karein.


Incorrect PIN par remaining attempts show karein.



Three incorrect attempts ke baad Account Blocked display karein.


````python
secret = 1234
for attempt in range(1,4):
  pin = int(input("Enter Pin : "))
  if pin == secret:
    print("Access Granted")
    break
  else:
    print("Icorrect Pin")

    if attempt < 3:
      print(f"Remaiing Attempts : {3- attempt}")
    else:
      print("Account Blocked")
````
