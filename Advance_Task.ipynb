## Task 11: Personal Expense Tracker

### Task Description

Aisa program banayein jo user ke daily expenses record kare aur spending summary generate kare.

### Expense Information

Har expense ke liye:

- Expense name
- Category
- Amount

### Categories

```
Food
Transport

```
Shopping
Bills
Other
```

### Requirements

1. User se total expenses input lein.
2. Har expense ko dictionary mein store karein.
3. Tamam dictionaries ko list mein save karein.
4. Total spending calculate karein.
5. Har category ki spending calculate karein.
6. Highest expense find karein.
7. User ka budget input lein.
8. Budget exceed hone par warning display karein.
9. Function use karke final report display karein.

````python
total_expenses = int(input("Enter How Many Expenses You Want To Add : "))
expenses_list = []


for i in range(total_expenses):
  expense_name = input(f"\nEnter Expense {i+1} name : ")
  category = input(f"Enter Category of {expense_name} : ")
  amount = float(input(f"Enter Amount of {expense_name} : "))

  # dic me store kar rahay (MOVED INSIDE LOOP)
  expense_items = {
      'name' : expense_name,
      'category' : category,
      'amount' : amount
  }

  # ab list me dic ko dal rahay hai (MOVED INSIDE LOOP)
  expenses_list.append(expense_items)

# Total spending calculate karein.
total_spending = 0
for expense in expenses_list:
  total_spending += expense['amount']


# Har category ki spending calculate karein.

category_spending = {
    "Food": 0,
    "Transport": 0,
    "Shopping": 0,
    "Bills": 0,
    "Other": 0
}

for category_expense in expenses_list:
  category = category_expense['category'].title()
  amount = category_expense['amount']

  if category in category_spending:
    category_spending[category] += amount

  else:
    category_spending['Other'] += amount

highest_expense = None

# Corrected logic to find the highest expense

for expense in expenses_list:
  # Check if highest_expense is None (first item) or if current expense is greater

  if highest_expense is None or expense['amount'] > highest_expense['amount'] :
    highest_expense = expense

# user_budget input lena

user_budget = int(input("\nEnter Your Budget : "))

# Budget exceed hone par warning display karein.

if  total_spending  > user_budget :
  print("Warning : Your Budget Is Exceed")

else:
  print("You Are Within The Budget")

# Function use karke final report display karein.

def final_report():

  print("\n---------- Expense Report ----------")
  print(f"Total Spending : {total_spending}")
  print("\n Category Spending")
  for category,amount in category_spending.items():
    print(f"{category} : {amount}")

  print("\nHighest Expense")
  # Added a check for highest_expense to handle cases where expenses_list might be empty
  if highest_expense:
    print(f"Name : {highest_expense['name']}")
    print(f"Amount : {highest_expense['amount']}")
  else:
    print("No expenses recorded.")

final_report()
````

## Task 12: Store Inventory Manager

### Task Description

Ek small shop ke products aur stock ko manage karne wala program banayein.

### Product Information

- Product name
- Price
- Available quantity

### Menu

```
1. Add Product
2. Sell Product
3. Restock Product
4. Search Product
5. Display Inventory
Exit
### Requirements

1. Products ko dictionary mein store karein.
2. New product add karne ka function banayein.
3. Product sell hone par stock reduce karein.
4. Insufficient stock par warning show karein.
5. Restocking par quantity increase karein.
6. Product name se search karein.
7. Low-stock products display karein.
8. Low stock ki condition quantity below `5` hogi.
9. Total inventory value calculate karein.

### Formula

Product Value = Price × Quantity



````python
products = {}


# 1. Add Product
def add_product():
    name = input("Enter product name: ").title()

    if name in products:
        print("Product already exists!")
        return

    price = float(input("Enter product price: "))
    quantity = int(input("Enter available quantity: "))

    products[name] = {
        "Price": price,
        "Quantity": quantity
    }

    print("Product added successfully!")


# 2. Sell Product
def sell_product():
    name = input("Enter product name: ").title()

    if name not in products:
        print("Product not found!")
        return

    quantity = int(input("Enter quantity to sell: "))

    if quantity > products[name]["Quantity"]:
        print("Insufficient stock!")
        return

    products[name]["Quantity"] -= quantity

    print("Product sold successfully!")


# 3. Restock Product
def restock_product():
    name = input("Enter product name: ").title()

    if name not in products:
        print("Product not found!")
        return

    quantity = int(input("Enter quantity to add: "))

    products[name]["Quantity"] += quantity

    print("Product restocked successfully!")


# 4. Search Product
def search_product():
    name = input("Enter product name to search: ").title()

    if name in products:
        print("\nProduct Found!")
        print("Product:", name)
        print("Price: Rs.", products[name]["Price"])
        print("Quantity:", products[name]["Quantity"])
    else:
        print("Product not found!")


# 5. Display Inventory
def display_inventory():
    if len(products) == 0:
        print("Inventory is empty!")
        return

    print("\n===== Inventory =====")

    for name, details in products.items():

        print("Product:", name)
        print("Price: Rs.", details["Price"])
        print("Quantity:", details["Quantity"])

        # Low stock check
        if details["Quantity"] < 5:
            print("Status: LOW STOCK")

        print("--------------------")


# 6. Total Inventory Value
def total_inventory_value():
    total = 0

    for name, details in products.items():
        value = details["Price"] * details["Quantity"]
        total += value

    print("Total Inventory Value: Rs.", total)


# Main Menu
while True:

    print("\n===== Store Inventory Manager =====")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Total Inventory Value")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        total_inventory_value()

    elif choice == "7":
        print("Program closed.")
        break

    else:
        print("Invalid choice!")
````

## Task 13: Library Book Management System

### Task Description

Ek small library ke books aur borrowing records manage karne wala program banayein.

### Book Information

- Book ID
- Book title
- Author
- Availability status

### Menu

```
1. Add Book
2. Display Books
3. Search Book
4. Borrow Book
5. Return Book
6. Exit
### Requirements

1. Har book ko dictionary mein store karein.
2. Tamam books ko list mein save karein.
3. Book ID unique honi chahiye.
4. Title se book search karein.
5. Sirf available book borrow ho sake.
6. Borrowed book ka status update karein.
7. Return karne par book dobara available karein.
8. Har menu operation ke liye separate function banayein.

````python
books = []


# 1. Add Book
def add_book():
    book_id = input("Enter Book ID: ")

    # Check duplicate Book ID
    for book in books:
        if book["Book ID"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "Book ID": book_id,
        "Title": title,
        "Author": author,
        "Available": True
    }

    books.append(book)

    print("Book added successfully!")


# 2. Display Books
def display_books():
    if len(books) == 0:
        print("No books available.")
        return

    print("\n===== Library Books =====")

    for book in books:
        print("Book ID:", book["Book ID"])
        print("Title:", book["Title"])
        print("Author:", book["Author"])

        if book["Available"]:
            print("Status: Available")
        else:
            print("Status: Borrowed")

        print("------------------------")


# 3. Search Book
def search_book():
    title = input("Enter book title to search: ").lower()

    for book in books:
        if book["Title"].lower() == title:
            print("\nBook Found!")
            print("Book ID:", book["Book ID"])
            print("Title:", book["Title"])
            print("Author:", book["Author"])

            if book["Available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            return

    print("Book not found!")


# 4. Borrow Book
def borrow_book():
    book_id = input("Enter Book ID to borrow: ")

    for book in books:

        if book["Book ID"] == book_id:

            if book["Available"]:
                book["Available"] = False
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed!")

            return

    print("Book ID not found!")


# 5. Return Book
def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in books:

        if book["Book ID"] == book_id:

            if book["Available"] == False:
                book["Available"] = True
                print("Book returned successfully!")
            else:
                print("This book is already available!")

            return

    print("Book ID not found!")


# Main Menu
while True:

    print("\n===== Library Book Management System =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("Program closed.")
        break

    else:
        print("Invalid choice!")
````

## Task 14: Restaurant Ordering System

### Task Description

Customer restaurant menu se multiple food items order kar sake aur program complete bill generate kare.

### Example Menu

```
menu= {"Burger":500,"Pizza":1200,"Pasta":800,"Drink":200
}
### Requirements

1. Restaurant menu display karein.
2. Customer se item name aur quantity input lein.
3. Check karein ke item menu mein available hai.
4. Multiple items order karne ki facility dein.
5. Har item ka subtotal calculate karein.
6. Order ko list of dictionaries mein store karein.
7. Total bill calculate karein.
8. Total bill par 13% tax calculate karein.
9. Rs. 3,000 ya above par 10% discount dein.
10. Function use karke formatted receipt display karein.

````python
menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

orders = []


# 1. Display Menu
def display_menu():
    print("\n===== Restaurant Menu =====")

    for item, price in menu.items():
        print(item, "Rs.", price)


# 2. Add Order
def add_order():
    item = input("Enter item name: ").title()

    # Check item available hai ya nahi
    if item not in menu:
        print("Item not available!")
        return

    quantity = int(input("Enter quantity: "))

    price = menu[item]
    subtotal = price * quantity

    order = {
        "Item": item,
        "Quantity": quantity,
        "Price": price,
        "Subtotal": subtotal
    }

    orders.append(order)

    print("Item added to order!")


# 3. Calculate Total
def calculate_total():
    total = 0

    for order in orders:
        total += order["Subtotal"]

    return total


# 4. Display Receipt
def display_receipt():

    if len(orders) == 0:
        print("No items ordered!")
        return

    print("\n========== RECEIPT ==========")

    for order in orders:
        print(
            order["Item"],
            "x",
            order["Quantity"],
            "=",
            "Rs.",
            order["Subtotal"]
        )

    total = calculate_total()

    # Tax 13%
    tax = total * 0.13

    # Discount
    discount = 0

    if total >= 3000:
        discount = total * 0.10

    final_bill = total + tax - discount

    print("-----------------------------")
    print("Subtotal: Rs.", total)
    print("Tax (13%): Rs.", tax)
    print("Discount: Rs.", discount)
    print("-----------------------------")
    print("Final Bill: Rs.", final_bill)
    print("=============================")


# Main Program
while True:

    print("\n===== Restaurant Ordering System =====")
    print("1. Display Menu")
    print("2. Add Order")
    print("3. Display Receipt")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        display_menu()

    elif choice == "2":
        add_order()

    elif choice == "3":
        display_receipt()

    elif choice == "4":
        print("Thank you for visiting!")
        break

    else:
        print("Invalid choice!")
````

## Task 15: Clinic Appointment Manager

### Task Description

Aisa program banayein jo clinic ke patients aur appointments manage kare.

### Patient Information

- Patient ID
- Patient name
- Age
- Disease/problem
- Appointment time
- Emergency status

### Requirements
1. Patient information input lein.
2. Har patient ko dictionary mein store karein.
3. Tamam patients ko list mein save karein.
4. Patient ID duplicate nahi honi chahiye.
5. Patient ID se record search karein.
6. Normal aur emergency patients ko identify karein.
7. Emergency patients ko list mein pehle display karein.
8. Appointment cancel karne ki facility dein.
9. Total normal aur emergency patients count karein.
10. Different operations ke liye separate functions banayein.
### Suggested Menu

```
1. Add Appointment
2. Search Patient
3. Display Appointments
4. Cancel Appointment
5. Display Clinic Summary
6. Exit


````python
patients = []


# 1. Add Appointment
def add_appointment():
    patient_id = input("Enter Patient ID: ")

    # Check duplicate ID
    for patient in patients:
        if patient["Patient ID"] == patient_id:
            print("Patient ID already exists!")
            return

    name = input("Enter Patient Name: ")
    age = int(input("Enter Age: "))
    disease = input("Enter Disease/Problem: ")
    appointment_time = input("Enter Appointment Time: ")
    emergency = input("Emergency? (yes/no): ").lower()

    if emergency == "yes":
        emergency_status = True
    else:
        emergency_status = False

    patient = {
        "Patient ID": patient_id,
        "Patient Name": name,
        "Age": age,
        "Disease": disease,
        "Appointment Time": appointment_time,
        "Emergency": emergency_status
    }

    patients.append(patient)

    print("Appointment added successfully!")


# 2. Search Patient
def search_patient():
    patient_id = input("Enter Patient ID to search: ")

    for patient in patients:
        if patient["Patient ID"] == patient_id:
            print("\nPatient Found:")
            print("Patient ID:", patient["Patient ID"])
            print("Name:", patient["Patient Name"])
            print("Age:", patient["Age"])
            print("Disease:", patient["Disease"])
            print("Appointment Time:", patient["Appointment Time"])

            if patient["Emergency"]:
                print("Emergency: Yes")
            else:
                print("Emergency: No")

            return

    print("Patient not found!")


# 3. Display Appointments
def display_appointments():
    if len(patients) == 0:
        print("No appointments available.")
        return

    # Emergency patients first
    print("\n--- Emergency Patients ---")

    for patient in patients:
        if patient["Emergency"]:
            print(
                patient["Patient ID"],
                "|",
                patient["Patient Name"],
                "|",
                patient["Appointment Time"]
            )

    print("\n--- Normal Patients ---")

    for patient in patients:
        if not patient["Emergency"]:
            print(
                patient["Patient ID"],
                "|",
                patient["Patient Name"],
                "|",
                patient["Appointment Time"]
            )


# 4. Cancel Appointment
def cancel_appointment():
    patient_id = input("Enter Patient ID to cancel: ")

    for patient in patients:
        if patient["Patient ID"] == patient_id:
            patients.remove(patient)
            print("Appointment cancelled successfully!")
            return

    print("Patient not found!")


# 5. Clinic Summary
def clinic_summary():
    emergency_count = 0
    normal_count = 0

    for patient in patients:
        if patient["Emergency"]:
            emergency_count += 1
        else:
            normal_count += 1

    print("\n--- Clinic Summary ---")
    print("Total Patients:", len(patients))
    print("Emergency Patients:", emergency_count)
    print("Normal Patients:", normal_count)


# Main Menu
while True:

    print("\n===== Clinic Appointment Manager =====")
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_appointment()

    elif choice == "2":
        search_patient()

    elif choice == "3":
        display_appointments()

    elif choice == "4":
        cancel_appointment()

    elif choice == "5":
        clinic_summary()

    elif choice == "6":
        print("Program closed.")
        break

    else:
        print("Invalid choice!")
````
